```python
import os
import numpy as np
import imageio

index = 133

lpips_image_path = f"../all_images/lpips/sr_{index}.png"
cpsnr_image_path = f"../all_images/cPSNR/sr_{index}.png"
hr_path = f"../all_images/hr/hr_{index}.png"

lpips_image = imageio.imread(lpips_image_path)
cpsnr_image = imageio.imread(cpsnr_image_path)
hr_image = imageio.imread(hr_path)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Convert images to grayscale if they are not already
if lpips_image.ndim == 3:
    lpips_image = np.mean(lpips_image, axis=2)  # Convert to grayscale

if cpsnr_image.ndim == 3:
    cpsnr_image = np.mean(cpsnr_image, axis=2)  # Convert to grayscale

if hr_image.ndim == 3:
    hr_image = np.mean(hr_image, axis=2)  # Convert to grayscale

# Compute the 2D Fourier Transform
f_transform_lpips = np.fft.fft2(lpips_image)
f_transform_cpsnr = np.fft.fft2(cpsnr_image)
f_transform_hr = np.fft.fft2(hr_image)

# Shift the zero frequency components to the center
f_shifted_lpips = np.fft.fftshift(f_transform_lpips)
f_shifted_cpsnr = np.fft.fftshift(f_transform_cpsnr)
f_shifted_hr = np.fft.fftshift(f_transform_hr)

# Compute the magnitude spectrum
magnitude_spectrum_lpips = np.log(np.abs(f_shifted_lpips) + 1)
magnitude_spectrum_cpsnr = np.log(np.abs(f_shifted_cpsnr) + 1)
magnitude_spectrum_hr = np.log(np.abs(f_shifted_hr) + 1)

# Plotting the images and their magnitude spectrums
plt.figure(figsize=(16, 12))  # Increase figure size to accommodate six subplots

# Original LPIPS Image
plt.subplot(231)
plt.imshow(lpips_image, cmap='gray')
plt.title('Original LPIPS Image')
plt.axis('off')

# Magnitude Spectrum of LPIPS Image
plt.subplot(234)
plt.imshow(magnitude_spectrum_lpips, cmap='gray')
plt.title('Magnitude Spectrum of LPIPS')
plt.axis('off')

# Original CPSNR Image
plt.subplot(232)
plt.imshow(cpsnr_image, cmap='gray')
plt.title('Original CPSNR Image')
plt.axis('off')

# Magnitude Spectrum of CPSNR Image
plt.subplot(235)
plt.imshow(magnitude_spectrum_cpsnr, cmap='gray')
plt.title('Magnitude Spectrum of CPSNR')
plt.axis('off')

# Original HR Image
plt.subplot(233)
plt.imshow(hr_image, cmap='gray')
plt.title('Original HR Image')
plt.axis('off')

# Magnitude Spectrum of HR Image
plt.subplot(236)
plt.imshow(magnitude_spectrum_hr, cmap='gray')
plt.title('Magnitude Spectrum of HR')
plt.axis('off')

plt.show()
```
